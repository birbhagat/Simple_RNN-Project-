```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent=[
    "Python is an easy-to-learn programming language.",
    "It is widely used in data science and AI.",
    "Python has a clean and readable syntax.",
    "It supports multiple programming paradigms.",
    "Libraries like NumPy and TensorFlow make Python powerful.",
    "Python has a massive and helpful developer community."

]
```

```python
sent
```

```python
## Define   The Vocabulary Size
voc_size=10000
```

```python
## One Hot Representation

one_hot_rep=[one_hot(words,voc_size)for words in sent]
one_hot_rep
```

```python
## Word Embedding Representation

from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential 
```

```python
import numpy as np
```

```python
## SENTENCE SHOULD BE OF SAME LENGTH
sent_length=8
Embedded_doc=pad_sequences(one_hot_rep,padding='pre',maxlen=sent_length)
print(Embedded_doc)
```

```python
## Feature_Representation
dim=10
```

```python
model=Sequential()
model.add(Embedding(voc_size,dim,input_length=sent_length))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(Embedded_doc)
```
